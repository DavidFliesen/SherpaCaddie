This section downloads Harry Vardon's *The Complete Golfer* from Project Gutenberg, removes boilerplate and front matter, then keeps paragraphs strongly related to confidence, patience, nerve, temperament, concentration, decision-making, and recovery. Equipment-heavy and mechanical passages are filtered out.

The resulting excerpts are appended only to the mental-game file, under a clearly labeled historical section. The Caddie Playbook instructs Sage to translate the ideas into modern language and never treat 1905 equipment, technique, or rules as current guidance.

```python
CANDIDATE_URLS = [
    "https://www.gutenberg.org/cache/epub/28107/pg28107.txt",
    "https://www.gutenberg.org/files/28107/28107-0.txt",
    "https://www.gutenberg.org/files/28107/28107.txt",
]

MENTAL_KEYWORDS = [
    "nerve", "nerves", "nervous", "confidence", "confident",
    "patience", "patient", "temper", "temperament", "calm",
    "steadiness", "steady", "anxious", "anxiety", "fear", "doubt",
    "hesitat", "courage", "philosoph", "worry", "despair",
    "despond", "determination", "resolve", "concentrat", "cautious",
    "caution", "bold", "risk", "pressing", "state of mind",
    "excitement", "disappoint", "recover", "self-control",
    "decision", "decide", "definite", "hopeless", "hurry", "haste",
]

# Strong signs that a passage is mainly obsolete equipment or detailed mechanics.
OBSOLETE_OR_MECHANICAL = [
    "gutta-percha", "gutty", "hickory", "brassie", "brassy", "cleek", "mashie",
    "niblick", "bulger", "socket", "whipping", "club-maker", "club maker",
    "wooden club", "iron club", "shaft", "lofting iron", "putting cleek",
    "stance diagram", "photograph shows", "illustration shows",
]

# These chapters contain the strongest material for learning, putting composure,
# general mental habits, competition, partnership, caddies, and reflection.
# Autobiography, equipment construction, course travelogues, the obsolete rules
# appendix, and the index are intentionally excluded before keyword filtering.
MENTAL_SOURCE_CHAPTERS = {"III", "XV", "XVI"}

IRRELEVANT_TOPICS = {
    "eyesight", "spectacles", "eye-glasses", "billiards", "shooting spectacles",
}

def fetch_vardon_text():
    headers = {"User-Agent": "Mozilla/5.0 (Sherpa-Caddie-Knowledge-Builder)"}
    errors = []
    for url in CANDIDATE_URLS:
        try:
            req = urllib.request.Request(url, headers=headers)
            with urllib.request.urlopen(req, timeout=25) as response:
                text = response.read().decode("utf-8", errors="replace")
            return text, url, errors
        except Exception as exc:
            errors.append(f"{url}: {exc}")
    return None, None, errors

def strip_gutenberg_boilerplate(text):
    start_match = re.search(
        r"\*\*\*\s*START OF (?:THIS|THE) PROJECT GUTENBERG EBOOK.*?\*\*\*",
        text,
        re.IGNORECASE,
    )
    end_match = re.search(
        r"\*\*\*\s*END OF (?:THIS|THE) PROJECT GUTENBERG EBOOK.*?\*\*\*",
        text,
        re.IGNORECASE,
    )
    start = start_match.end() if start_match else 0
    end = end_match.start() if end_match else len(text)
    body = text[start:end]
    anchor = body.find("I have sometimes heard good golfers")
    if anchor == -1:
        anchor = body.find("CHAPTER I")
    body = body[anchor:] if anchor != -1 else body
    body = re.split(r"(?m)^APPENDIX\s*$", body, maxsplit=1)[0]
    return body

def select_mental_game_chapters(body):
    matches = list(re.finditer(r"(?m)^CHAPTER\s+([IVXLCDM]+)\s*$", body))
    selected = []
    for index, match in enumerate(matches):
        chapter = match.group(1)
        end = matches[index + 1].start() if index + 1 < len(matches) else len(body)
        if chapter in MENTAL_SOURCE_CHAPTERS:
            selected.append(body[match.end():end])
    return "\n\n".join(selected)

def clean_source_paragraph(paragraph):
    paragraph = re.sub(r"\[Pg\s*\d+\]", "", paragraph)
    paragraph = re.sub(r"\s+", " ", paragraph).strip()
    return paragraph

def mental_score(paragraph):
    lower = paragraph.lower()
    score = sum(lower.count(keyword) for keyword in MENTAL_KEYWORDS)
    penalty = sum(2 for term in OBSOLETE_OR_MECHANICAL if term in lower)
    return score - penalty

def extract_mental_game_paragraphs(raw_text):
    body = strip_gutenberg_boilerplate(raw_text)
    body = select_mental_game_chapters(body)
    raw_paragraphs = re.split(r"\n\s*\n", body)
    candidates = []
    seen = set()

    for source_index, raw in enumerate(raw_paragraphs):
        paragraph = clean_source_paragraph(raw)
        if not paragraph or len(paragraph) < 100 or len(paragraph) > 4000 or paragraph.isupper():
            continue
        if paragraph.count("—") > 6 or paragraph.count("--") > 6:
            continue
        if len(re.findall(r";\s*\d+", paragraph)) >= 2:
            continue
        if any(topic in paragraph.lower() for topic in IRRELEVANT_TOPICS):
            continue
        key = re.sub(r"\W+", " ", paragraph.lower()).strip()
        if key in seen:
            continue
        seen.add(key)
        score = mental_score(paragraph)
        obsolete_hits = sum(paragraph.lower().count(term) for term in OBSOLETE_OR_MECHANICAL)
        if obsolete_hits >= 2 and score < 6:
            continue
        if score >= MIN_MENTAL_SCORE:
            candidates.append((source_index, score, paragraph))

    # If the filter finds more than the configured cap, keep the strongest material,
    # then restore book order so the historical section reads coherently.
    if len(candidates) > MAX_HISTORIC_PARAGRAPHS:
        candidates = sorted(candidates, key=lambda item: (-item[1], item[0]))[:MAX_HISTORIC_PARAGRAPHS]
    candidates.sort(key=lambda item: item[0])
    return [paragraph for _, _, paragraph in candidates]

vardon_paragraphs = []
vardon_url = None
fetch_errors = []

if INCLUDE_VARDON_MENTAL_GAME:
    raw_vardon, vardon_url, fetch_errors = fetch_vardon_text()
    if raw_vardon:
        vardon_paragraphs = extract_mental_game_paragraphs(raw_vardon)
        print(f"Selected {len(vardon_paragraphs)} mental-game paragraphs from {vardon_url}")
    else:
        print("Vardon download was unavailable. The five complete core files will still be generated.")
        for error in fetch_errors:
            print(" -", error)
else:
    print("Vardon enrichment disabled; generating the complete Sherpa-authored corpus.")
```

## 3. Assemble the exact five documents

Each file receives a common philosophy and live-data safeguard. The Mental Game file receives the complete modern Sherpa mental framework plus the filtered historical enrichment, when available.

```python
def render_document(title, core_text, source_note):
    return "\n\n".join([
        COMMON_HEADER,
        f"DOCUMENT TITLE\n{title}",
        core_text.strip(),
        f"SOURCE AND USE NOTE\n{source_note.strip()}",
    ]).strip() + "\n"

historical_section = ""
if vardon_paragraphs:
    numbered = []
    for index, paragraph in enumerate(vardon_paragraphs, start=1):
        numbered.append(f"HISTORICAL MENTAL-GAME PASSAGE {index}\n{paragraph}")
    historical_section = "\n\n".join([
        "HISTORICAL MENTAL-GAME WISDOM — HARRY VARDON, 1905",
        (
            "The passages below come from the public-domain book The Complete Golfer. "
            "They are retained for observations about patience, nerve, confidence, temperament, "
            "judgment, and recovery. Sage should convey the useful idea in modern language. "
            "Any period equipment, swing instruction, course condition, or rule is historical—not current guidance."
        ),
        *numbered,
    ])
else:
    historical_section = dedent("""
    HISTORICAL MENTAL-GAME WISDOM
    Public-domain enrichment was not downloaded during this run. The complete modern Sherpa mental-game framework above remains available. Re-run this notebook with internet access to append the filtered Harry Vardon material.
    """).strip()

documents = {
    "Sherpa_Caddie_Golf_Fundamentals.txt": render_document(
        "Golf Fundamentals",
        GOLF_FUNDAMENTALS,
        "Original Sherpa Caddie educational material. General instruction only; live recommendations must use Sherpa tools.",
    ),
    "Sherpa_Caddie_Course_Strategy.txt": render_document(
        "Course Strategy",
        COURSE_STRATEGY,
        "Original Sherpa Caddie course-management material. Current distances, weather, clubs, and round state must come from Sherpa tools.",
    ),
    "Sherpa_Caddie_Mental_Game.txt": render_document(
        "Mental Game",
        MENTAL_GAME + "\n\n" + historical_section,
        (
            "Modern sections are original Sherpa Caddie material. Historical passages, when present, "
            "are filtered from Harry Vardon's The Complete Golfer (1905), Project Gutenberg ebook #28107, "
            f"public domain in the United States. Download source used: {vardon_url or 'not available during this run'}."
        ),
    ),
    "Sherpa_Caddie_Rules_Etiquette_and_Golfer_Guide.txt": render_document(
        "Rules, Etiquette, and Golfer Guide",
        RULES_ETIQUETTE_GUIDE,
        "Original general orientation. It intentionally defers exact or competitive rulings to current official rules, local rules, and qualified officials.",
    ),
    "Sherpa_Caddie_Caddie_Playbook.txt": render_document(
        "Caddie Playbook",
        CADDIE_PLAYBOOK,
        "Original Sherpa Caddie behavior and tool-use guidance for Sage. Agent-level instructions should reinforce these same priorities.",
    ),
}

assert list(documents.keys()) == EXPECTED_FILES
print("Assembled:")
for name, text in documents.items():
    print(f" - {name}: {len(text):,} characters")
```

## 4. Write and validate the D-ID upload folder

The dedicated output directory is rebuilt so stale chunks or manifests from the older notebook cannot be uploaded accidentally. Validation confirms that the folder contains exactly the expected five nonempty TXT files and that every file is below D-ID's displayed 20 MB per-file limit.

```python
# Safety guard: only rebuild the one dedicated, explicitly named output directory.
if OUTPUT_DIR.name != "Sherpa_Caddie_DID_Knowledge":
    raise RuntimeError("Refusing to rebuild an unexpected output directory.")

if OUTPUT_DIR.exists():
    shutil.rmtree(OUTPUT_DIR)
OUTPUT_DIR.mkdir(parents=True)

for filename in EXPECTED_FILES:
    (OUTPUT_DIR / filename).write_text(documents[filename], encoding="utf-8")

actual_files = sorted(path.name for path in OUTPUT_DIR.iterdir() if path.is_file())
expected_sorted = sorted(EXPECTED_FILES)

if actual_files != expected_sorted:
    raise RuntimeError(f"Output mismatch. Expected {expected_sorted}; found {actual_files}")

validation_rows = []
for filename in EXPECTED_FILES:
    path = OUTPUT_DIR / filename
    size_bytes = path.stat().st_size
    text = path.read_text(encoding="utf-8")
    if not text.strip():
        raise RuntimeError(f"{filename} is empty")
    if size_bytes > MAX_FILE_BYTES:
        raise RuntimeError(f"{filename} exceeds the 20 MB D-ID per-file limit")
    validation_rows.append({
        "file": filename,
        "bytes": size_bytes,
        "characters": len(text),
        "words": len(text.split()),
        "within_20_MB": True,
    })

print("VALIDATION PASSED — exactly five D-ID-ready TXT files were generated.\n")
for row in validation_rows:
    print(f"{row['file']}\n  {row['words']:,} words | {row['characters']:,} characters | {row['bytes']:,} bytes")
```
